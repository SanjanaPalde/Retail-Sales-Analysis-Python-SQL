## 4. Sales Analysis

```python
#We've now completed:
#💰 Total Revenue → ₹56,37,637.05
#📊 Revenue by Category
#🏙️ Revenue by City
#💳 Revenue by Payment Method
#📦 Best-selling Products
#📅 Monthly Revenue
#📋 Order Status Analysis
#💰 Average Order Value (AOV)
#   highest-value order.
#   Top Customers by Revenue
```

```python
    #analyze Revenue
df['Total_Amount'].sum()
```

```python
    #Revenue by Category  display only last expression
df.groupby("Category")["Total_Amount"].sum().sort_values(ascending=False)
# df.groupby("City")["Total_Amount"].sum().sort_values(ascending=False)
# df.groupby("Payment_Method")["Total_Amount"].sum().sort_values(ascending=False)("Product_Name")["Quantity"].sum().sort_values(ascending=False)
```

## 5. Time & Order Analysis

```python
# Monthly Revenue
df.groupby(df["Order_Date"].dt.month_name())["Total_Amount"].sum().sort_values(ascending=False)
```

```python
# Order Status Analysis
df["Order_Status"].value_counts()
```

```python
# Average Order Value
print("avg",df["Total_Amount"].mean())
```

```python
# Highest Order Value
df["Total_Amount"].max()
```

## 6. Customer Analysis

```python
#  This converts:KAVYA JOSHI → Kavya Joshi
df["Clean_Customer_Name"] = df["Customer_Name"].str.title()
```

```python
# Top Customers by Revenue
df.groupby(["Correct_cust_id", "Clean_Customer_Name"])["Total_Amount"].sum().sort_values(ascending=False)
```
